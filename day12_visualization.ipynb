## Student Performance Visualization

This notebook uses Matplotlib and Seaborn to visualize student performance.

### Visualizations
- Bar chart
- Histogram
- Scatter plot
- Box plot
- Seaborn bar plot
- Correlation heatmap

### Goal
Understand patterns, distributions, relationships, and possible outliers in the student dataset.

```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
```

```python
data = {
    "Student": ["Arun", "Bala", "Kumar", "Divya", "Ravi", "Priya", "Suresh", "Meena"],
    "StudyHours": [1, 2, 3, 4, 5, 6, 2, 4],
    "Marks": [45, 50, 58, 65, 72, 85, 52, 68]
}

df = pd.DataFrame(data)

df
```

```python
plt.bar(df["Student"], df["Marks"])

plt.xlabel("Student")
plt.ylabel("Marks")
plt.title("Student Marks")

plt.show()
```

```python
plt.hist(df["Marks"], bins=5)

plt.xlabel("Marks")
plt.ylabel("Number of Students")
plt.title("Distribution of Student Marks")

plt.show()
```

```python
plt.scatter(df["StudyHours"], df["Marks"])

plt.xlabel("Study Hours")
plt.ylabel("Marks")
plt.title("Study Hours vs Marks")

plt.show()
```

```python
plt.boxplot(df["Marks"])

plt.ylabel("Marks")
plt.title("Student Marks Box Plot")

plt.show()
```

```python
sns.barplot(x="Student", y="Marks", data=df)

plt.xlabel("Student")
plt.ylabel("Marks")
plt.title("Student Marks using Seaborn")

plt.show()
```

```python
correlation = df[["StudyHours", "Marks"]].corr()

print(correlation)
```

```python
sns.heatmap(correlation, annot=True)

plt.title("Study Hours and Marks Correlation")

plt.show()
```

## Observations

1. Students who studied more hours generally scored higher marks.
2. The highest mark is 85 and the lowest mark is 45.
3. Priya scored the highest mark with 85 marks.
4. The marks do not show an obvious extreme outlier.
5. Study hours and marks have a strong positive relationship.
