# Day 15 - Machine Learning Fundamentals

## What is Machine Learning?
ML allows computers to learn patterns from data and make predictions or decisions.

## Features
Input information used by the model.

## Labels
The answer/target the model learns to predict.

## Training
Model learns patterns from training data.

## Validation
Used to evaluate and tune the model during development.

## Testing
Used for final evaluation on unseen data.

## Supervised Learning
Learning from data with labels.
Examples: regression and classification.

## Unsupervised Learning
Finding patterns or groups without predefined labels.
Example: clustering.

## ML Pipeline
Training → Validation → Testing → Prediction
