# Day 11 - Data Cleaning

## Objective

Clean a messy student dataset using Pandas.

### Cleaning steps
- Inspect the dataset
- Find missing values
- Remove duplicate rows
- Fill missing values
- Fix incorrect values
- Convert data types
- Check the final cleaned dataset

```python
import pandas as pd
```

```python
data = {
    "Name": ["Arun", "Bala", "Kumar", "Arun", "Divya"],
    "Age": ["21", "22", None, "21", "twenty"],
    "Marks": [85, 72, 91, 85, None]
}

df = pd.DataFrame(data)

df
```

## 1. Inspect the Messy Data

Before cleaning, we inspect the dataset to find problems.

```python
print("Data Types:")
print(df.dtypes)

print("\nMissing Values:")
print(df.isnull().sum())

print("\nDuplicate Rows:")
print(df.duplicated())
```

## 2. Remove Duplicate Rows

Duplicate rows can cause incorrect results during data analysis and machine learning.

```python
df = df.drop_duplicates()

print("After Removing Duplicates:")
print(df)
```

## 3. Handle Missing Age

The Age column contains one missing value.
We will fill it with 20 for this practice dataset.

```python
df["Age"] = df["Age"].fillna("20")

print("After Filling Missing Age:")
print(df)
```

## 4. Fix Incorrect Age Values

The Age column contains `"twenty"` as text.
We will replace it with `"20"` and then convert the column to integers.

```python
df["Age"] = df["Age"].replace("twenty", "20")

df["Age"] = df["Age"].astype(int)

print("After Cleaning Age:")
print(df)

print("\nData Types:")
print(df.dtypes)
```

## 5. Handle Missing Marks

The Marks column contains one missing value.
We will fill it using the mean of the available marks.

```python
mean_marks = df["Marks"].mean()

print("Mean Marks:", mean_marks)

df["Marks"] = df["Marks"].fillna(mean_marks)

print("\nAfter Cleaning Marks:")
print(df)
```

## 6. Final Verification

After cleaning, we verify that there are no missing values or duplicate rows and that the data types are correct.

```python
print("Final Data:")
print(df)

print("\nFinal Data Types:")
print(df.dtypes)

print("\nFinal Missing Values:")
print(df.isnull().sum())

print("\nFinal Duplicate Rows:")
print(df.duplicated().sum())
```

## 7. Summary

The messy student dataset was successfully cleaned.

### Cleaning performed
- Removed duplicate rows using `drop_duplicates()`
- Filled missing Age using `fillna()`
- Replaced the incorrect `"twenty"` value
- Converted Age from string to integer using `astype(int)`
- Filled missing Marks using the mean
- Verified that there are no missing values
- Verified that there are no duplicate rows

### Final Result

The dataset is now ready for further data analysis and machine learning.
