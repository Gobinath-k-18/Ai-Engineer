## 1. What is SQL?

SQL stands for Structured Query Language.

SQL is used to:
- Store data
- Read data
- Update data
- Delete data
- Filter and analyze data

A relational database stores data in tables.

- Rows = records
- Columns = fields

````python
import sqlite3

connection = sqlite3.connect("students.db")
cursor = connection.cursor()

print("Connected to database!")
````

## 2. Create a Table

We create a `students` table with:

- id
- name
- age
- marks

````python
cursor.execute("""
CREATE TABLE IF NOT EXISTS students (
    id INTEGER PRIMARY KEY,
    name TEXT,
    age INTEGER,
    marks INTEGER
)
""")

connection.commit()

print("Students table created!")
````

## 1. What is SQL?

- Store data
- Read data
- Update data

- Filter and analyze data

A relational database stores data in tables.

- Rows = records

````python
import sqlite3

connection = sqlite3.connect("students.db")
cursor = connection.cursor()

print("Connected to database!")
````

## 2. Create a Table

We create a `students` table with:

- `id` → student ID
- `name` → student name
- `age` → student age
- `marks` → student marks

````python
cursor.execute("""
CREATE TABLE IF NOT EXISTS students (
    id INTEGER PRIMARY KEY,
    name TEXT,
    age INTEGER,
    marks INTEGER
)
""")

connection.commit()

print("Students table created!")
````

## 3. SELECT — Reading Data

`SELECT` is used to read data from a table.

```sql
SELECT * FROM students;

````python
cursor.execute("SELECT * FROM students")

rows = cursor.fetchall()

for row in rows:
    print(row)
````

## 4. SELECT Specific Columns

We don't always need every column.

We can select only the columns we want.

```sql
SELECT name, marks
FROM students;

````python
cursor.execute("SELECT name, marks FROM students")

rows = cursor.fetchall()

for row in rows:
    print(row)
````
